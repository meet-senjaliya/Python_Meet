The cost categories applied to a **$9.00 microcontroller** are:
```
- Engineering: $1.35  
- Manufacturing: $3.60  
- Sales: $2.25  
- Profit: $1.80
```
Create a **pie chart** showing the cost breakdown as **different sized pieces**.

````python
import matplotlib.pyplot as plt
categories = ['engg','mfg','sales','profit']
costs = [1.35,3.6,2.25,1.8]
totalcost = sum(costs)
percentages = [cost/totalcost*100 for cost in costs]
print(percentages)
plt.pie(percentages,
       labels = categories,
       autopct = "%1.1f%%")
plt.show()
````

## **711. Streamlit Profile App**

Create a **Streamlit app** where the user can:

* Enter **Name** (text input)
* Select **Age** (slider: 10–100)
* Choose **Gender** (radio: Male, Female, Other)
* Select multiple **Hobbies** (multiselect)
* Upload a **profile picture** (file uploader for images)

When the user clicks **Submit**:

* Display the profile details
* Show the uploaded image

---

````python
%%writefile pbstreamlit.py

################### PB 711 #########################
import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
st.set_page_config(page_title="Streamlit Example of PB",layout='wide')

name = st.text_input("Enter name")
age = st.slider("select age",10,100)
gender = st.radio("Choose Gender",['male','female','other'])
hobbies = st.multiselect(
"Select Hobbies",['Reading','Sports','music','travel','games','cooking'])
photo = st.file_uploader("Upload pofile pic",type = ['jpg','png','jpeg'])

if st.button("Submit Profile"):
    st.subheader("Profile Details")
    st.write("Name -",name)
    st.write("Age -",age)
    st.write("Gender -",gender)
    st.write("hobbies -",", ".join(hobbies)) 
    if photo:
        st.image(photo,caption='profile pic',width=200)
````

## **712. Streamlit Vaccination Percentage App**

Build a **Streamlit app** with the following features:

* Sidebar to select a **country** from a dropdown
  *(India, USA, UK, Canada)*
* Number input for **Total Population**
* Number input for **Vaccinated People**
* A **button** to calculate and display the **vaccination percentage**
* Display result using:

  * **Progress bar**
  * **Success / Warning message** depending on whether vaccination rate is above **70%**

````python
%%writefile pbstreamlit2.py

import streamlit as st
import numpy as np
st.header("PB 712 VACCINATION APP")
with st.sidebar:
    country = st.selectbox("Select Country",['India','USA','UK','Canada'])
    totalpopulation = st.number_input("Total population",min_value=1)
    vaccinatedpeople = st.number_input("Vaccinated People",min_value=0)
    
if st.button("Calculete vaccination %"):
    percentage = (vaccinatedpeople/totalpopulation)*100
    st.write(f"vaccination rate in {country}: {percentage:.2f}%")
    
    st.progress(min(int(percentage),100))
    if percentage >= 70:
        st.success("Good coverage of vaccination")
    else:
        st.warning("Vaccination cover is poor")
````

## **713. Streamlit Marks Calculator App**

Create a **Streamlit app** where the user can input marks of **5 subjects** using `number_input` arranged in **columns**.

Add a button to calculate and display:

* **Total marks**
* **Average marks**
* **Division** (First / Second / Fail based on average)

Display all results inside an **expander** section.

---

````python
%%writefile pb713.py

import streamlit as st
st.header("PB programm - 713")
c1,c2,c3,c4,c5 = st.columns(5)

m1 = st.number_input("FCSP-1: ",0,100,0)
m2 = st.number_input("FSD-1: ",0,100,0)
m3 = st.number_input("DE: ",0,100,0)
m4 = st.number_input("PS: ",0,100,0)
m5 = st.number_input("ETC: ",0,100,0)

if st.button("Calculate Marks"):
    total = m1 + m2 + m3 + m4 + m5
    average = total / 5
    if average >= 60:
        division = "First Division"
    elif average >= 40:
        division = "Second Division"
    else:
        division = "Fail"
    with st.expander("Result"):
        st.write("Total Marks   : ",total)
        st.write("Average Marks : ",average)
        st.write("Division      : ",division)
````

## **714. Streamlit BMI Calculator App**

Create a **BMI Calculator App** with user inputs:

* **Weight (kg)** – number input
* **Height (cm)** – number input

On button click, calculate:

```
BMI = weight / (height/100)^2
```

Display:

* **BMI value**
* **Health category** (Underweight, Normal, Overweight, Obese)
* Colored messages using:

  * `st.success()`
  * `st.warning()`
  * `st.error()`

````python
%%writefile pb714.py

import streamlit as st
st.header("PB programm - 714")
weight = st.number_input("Weight (kg) : ",min_value=1.0)
height = st.number_input("height (cm) : ",min_value=1.0)

if st.button("calculate BMI : "):
    bmi = weight / ((height / 100)**2)
    st.write(f"BMI VALUE : {bmi:.2f}")
    
    if bmi < 18.5:
        st.warning("Underweight - Eat more (balanced)")
    elif bmi < 25.0:
        st.success("Normal Weight - Keep It Up")
    elif bmi < 30:
        st.warning("Overweight - Cut down on overeating , Do more exercise")
    else:
        st.error("Obese")
````

## **715. Streamlit + Matplotlib Integration App**

Create a Streamlit app that:

* Takes number input **n** (number of random points)
* Generates **n random values** for `x` and `y`
* Plots them using **Matplotlib scatter plot**
* Displays the plot using `st.pyplot()`

---

````python
%%writefile pb715.py

import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
st.header("PB programm - 715")
n = st.number_input("Enter number of points to plot : ",min_value=10,step=10)

if st.button("Generate scatter plot"):
    x = np.random.rand(n)
    y = np.random.rand(n)
    
    plt.figure(figsize=(6,4))
    plt.scatter(x,y)
    plt.xlabel("X values")
    plt.ylabel("Y values")
    plt.title("Random Scatter Plot")
    
    st.pyplot(plt)
````

- Attribute = Two attribute to store Rs and Paise
- Set Value method = to set the rupee and pise values to money object
- `Addition` = overload + operator to add two money objects
- `Substraction` = overload - operator to subtract two money objects
- comparison - overload > operator to compare two money objects
- multiplication = overload * operator to multiply money object.
- Display method to print value in format - Rs X and Y paise.

**TEST CAE**
```
MONEY OBJECT 1 = m1 Rs 120 and 30 paise
MONEY OBJECT 2 = m1 Rs 80 and 80 paise
m1 + m2 = Rs 201 and 10 paise
m1 - m2 = Rs 30 and 50 Paise
if m1 = Rs 49 and 99 Paise,
m1 * 3 = Rs 149 and 97 paise

```

````python
class Money:
    def __init(self):
        self.R = 0
        self.P = 0
    def setvalue(self,R,P):
        self.R = R
        self.P = P
    def __add__(self,M):
        Temp = Money()
        Temp.R = self.R + M.R
        Temp.P = self.P + M.P
        if Temp.P >= 100:
            Temp.P -= 100
            Temp.R += 1
        return Temp
    def __sub__(self,M):
        Temp = Money()
        Temp.R = self.R - M.R
        Temp.P = self.P - M.P
        if Temp.P < 0:
            Temp.P += 100
            Temp.R -= 1
        return Temp
    def __gt__(self,M):
        return self.R > M.R
    def __mul__(self,n):
        Temp = self.R + (self.P)/100
        Temp = Temp * n
        print(f"{Temp:.2f}")
    def display(self):
        print("Rs",self.R,'and',self.P,'Paise')
````

````python
m1.setvalue(120,30)
m2.setvalue(80,80)
````
