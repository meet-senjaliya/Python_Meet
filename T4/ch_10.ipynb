## SUBPLOTS

````python
x = [1,2,3,4]
y = [10,20,30,40]
print(x,y)
````

````python
plt.subplot(1,2,1)
plt.plot(x,y,color="b")
plt.title("x vs y")

plt.subplot(1,2,2)
plt.plot(y,x,color="orange")
plt.title("y vs x")
plt.show()
````

## Scatter plot

````python
plt.scatter(x,y,
           color = 'r',
           s = 150,
           alpha = 0.6)
plt.grid()
plt.show()
````

## BARPLOT - vertical

````python
plt.bar(x,y,
       color = 'g',
       width = 0.5)
plt.show()
````

## BARPLOT - horizontal

````python
plt.barh(x,y,
        color = 'y',
        height = 0.6)
plt.show()
````

## HISTOGRAM

````python
data = [10,20,20,30,30,40]
plt.hist(data,
        bins = 4,
        orientation='vertical',
        align='mid')
plt.show()
````

## PIE CHART

````python
labels = ['A','B','C']
sizes = [40,35,25]
colors = ['r','y','g']
plt.pie(sizes,
       labels=labels,
       colors = colors,
       startangle = 90,
       autopct="%1.2f%%")
plt.show()
````

````python
import matplotlib.pyplot as plt
labels = ['A','B','C']
sizes = [40,35,25]
colors = ['r','y','g']
plt.pie(sizes,
       labels=labels,
       colors = colors,
       startangle = 90,
       autopct="%1.2f%%",
       explode = [0,0.1,0])
plt.show()
````

## AREAPLOT/FILL_BETWEEN AND STACKPLOT

````python
import matplotlib.pyplot as plt
import numpy as np

x = [1,2,3,4,5]
y = [2,4,3,8,7]

plt.fill_between(x,y)

plt.plot(x,y,
        color = 'r',
        linewidth = 3,
        label = "A Line")

plt.legend()
plt.show()
````

## STACKPLOT

````python
import matplotlib.pyplot as plt
import numpy as np

x = [1,2,3,4,5]
y = [1,4,6,8,9]
y1 = [2,2,7,10,12]
y2 = [2,8,5,10,6]

plt.stackplot(x,y,y1,y2,labels = ['A','B','C'])

plt.legend(loc = 'upper left')
plt.show()
````
